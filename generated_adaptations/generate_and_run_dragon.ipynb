## Run generated

```python
import subprocess
import os
import sys
from pathlib import Path
import textwrap
```

```python
dragon_configs = ["dragon/configs/default.yaml", "generated_adaptations/configs/generated.yaml"]
dragon_variants = [f"{llm}_{variant}_1" for llm in ("4o", "o3") for variant in ("default", "sd1", "sd2", "strategy")]
```

```python
dragon_variants = [f"{llm}_{variant}_1" for llm in ("4o", "o3") for variant in ("sd2",)]
```

### Prepare configuration files

```python
# prepare configuration files
for variant in dragon_variants:
    variant_path = Path(f"generated_adaptations/configs/dragon/{variant}.yaml")
    if not variant_path.exists():
        variant_path.parent.mkdir(parents=True, exist_ok=True)
        variant_path.write_text(f"""name: {variant}
log_dir.append: /{variant}
adaptation_name: generated_adaptations.dragon.{variant}.SmartAdaptation
adaptation_params: {{}}""")
```

```python
# prepare python files
for variant in dragon_variants:
    variant_path = Path(f"generated_adaptations/dragon/{variant}.py")
    if not variant_path.exists():
        variant_path.parent.mkdir(parents=True, exist_ok=True)
        variant_path.write_text("")
```

### Run

```python
def run(configs, repeats=10, start=1):
    print(configs)
    hps = []
    for repeat in range(repeats):
        print(f"  Run #{repeat + start}/{repeats + start - 1}")

        run_args = [sys.executable, "main.py", *configs, "-s", str(repeat + start), "-e", str(repeat + start)]
        # if repeat % 10 == 0:
        #     run_args.append("--animation")

        # disable TF errors
        env = dict(os.environ, TF_CPP_MIN_LOG_LEVEL="3")

        result = subprocess.run(run_args, capture_output=True, env=env)
        stdout = result.stdout.decode("utf-8")
        stderr = result.stderr.decode("utf-8")

        if stderr:
            print(textwrap.indent(stderr, "    "))
        else:
            hp = stdout.partition("dragon_hp: ")[2].partition("\n")[0]
            print(f"    Dragon HP: {hp}")
            hps.append(int(hp))

    if len(hps) > 0:
        avg_hp = sum(hps) / len(hps)
        print(f"Average HP: {avg_hp:.1f}")
    if repeats != len(hps):
        print(f"Errors: {repeats - len(hps)}")
    print()
```

```python
for variant in dragon_variants:
    configs = dragon_configs + [f"generated_adaptations/configs/dragon/{variant}.yaml"]
    run(configs, repeats=3)
```
