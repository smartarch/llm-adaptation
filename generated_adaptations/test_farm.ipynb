To run the tests from the terminal, use `pytest generated_adaptations/tests` from the project root directory (`llm-adaptation`).
It might be necessary to add the root directory to the Python path for the imports to work:
Windows: `set PYTHONPATH=%PYTHONPATH%;.`,
Linux: `PYTHONPATH=$PYTHONPATH:.`

Options: `-q` (less verbose output, we could also use `--no-header` instead to remove only platform, rootdir, plugins info), `--tb=no` (remove tracebacks, we could possibly also use `--tb=short` or `--tb=line` as information is in the summary are trimmed if the terminal is not wide enough).

Example: `pytest generated_adaptations/tests -q --tb=no --example=farm --adaptation_name=4o_sd2_4`

```python
def test(adaptation_name, example="farm"):
    return pytest.main(["generated_adaptations/tests", "-q", "--tb=no", f"--example={example}", f"--adaptation_name={adaptation_name}"])
```

```python
implementations = list(Path("generated_adaptations/farm").glob("*_v2.py"))
implementations
```

```python
failed = []
for implementation in implementations:
    print(f"Testing {implementation.name}")
    result = test(implementation.stem)
    if result != 0:
        failed.append(implementation.stem)
failed
```
